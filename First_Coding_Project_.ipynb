```python
import pandas as pd
data = {"Student" : ["Ali", "Ahmed", "Sara", "Ayesha", "Usman"],
"Math" : [78, 65, 92, 88, 55],
"English" : [72, 80, 95, 85, 60,],
"Science" : [85, 70, 90, 92, 58]}
df = pd.DataFrame(data)
df
```

```python
from google.colab import sheets
sheet = sheets.InteractiveSheet(df=df)
```

```python
df["Average"] = df [["Math","English", "Science"]].mean(axis=1)
print (df)
```

```python
top_student = df.loc [df["Average"].idxmax()]
print("Top Student:", top_student["Student"])
print("Average:", top_student["Average"])
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,5))
plt.bar(df["Student"], df["Average"])
plt.xlabel("Student")
plt.ylabel("Average")
plt.title("Average Marks of Students")
plt.ylim(0,100)
plt.show()
```

```python
subject_avg = df[["Math", "English", "Science"]].mean()
print("Best Subject:", subject_avg.idxmax())
print("Higher Average:", subject_avg.max())
```

```python
import matplotlib.pyplot as plt
subject_avg.plot(kind="bar")
plt.title("Subject-wise Average Performance")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.ylim(0,100)
plt.show()
```
